Atividade 4 — Dataclass

Crie uma dataclass chamada Aluno com:

nome
matricula
nota1
nota2
Crie um método chamado media() que calcule a média das duas notas.

Exemplo:

aluno = Aluno(
    "Maria",
    12345,
    8.0,
    9.0
)

print(aluno.media())

````python
from dataclasses import dataclass

@dataclass
class Aluno:
    nome: str
    matricula: int
    idade: int
    curso: str
    nota1: float
    nota2: float

    def media(self):
        return (self.nota1 + self.nota2) / 2

# Criando o objeto aluno passando os valores na ordem correta
aluno = Aluno(
    "Thales",
    12347,
    19,
    "computacao",
    9.0,
    9.0
)

# Testando o método media
print(aluno.media())
````

Atividade 5 — Lista de objetos

Utilizando a dataclass Aluno, cadastre cinco alunos.

Depois:

calcule a média de cada aluno;
exiba os alunos aprovados;
identifique o aluno com maior média.
Considere aprovação:

média >= 7.0


````python
#atividade 5
from dataclasses import dataclass

@dataclass
class Aluno:
    nome: str
    matricula: int
    idade: int
    curso: str
    nota1: float
    nota2: float

    def media(self):
        return (self.nota1 + self.nota2) / 2

# 1. Cadastrando os cinco alunos em uma lista
lista_alunos = [
    Aluno("Thales", 12347, 19, "Computação", 9.0, 9.0),
    Aluno("Maria", 12345, 20, "Engenharia", 8.0, 5.0),
    Aluno("João", 12348, 21, "Física", 4.0, 6.5),
    Aluno("Ana", 12349, 19, "Matemática", 10.0, 9.5),
    Aluno("Lucas", 12350, 22, "Computação", 7.0, 7.0)
]

# Variáveis para identificar o aluno com a maior média
aluno_maior_media = None
maior_media = -1.0

print("--- Médias de Cada Aluno ---")
for aluno in lista_alunos:
    m = aluno.media()
    print(f"Aluno: {aluno.nome} | Média: {m:.1f}")


    if m > maior_media:
        maior_media = m
        aluno_maior_media = aluno

print("\n--- Alunos Aprovados (Média >= 7.0) ---")
for aluno in lista_alunos:
    if aluno.media() >= 7.0:
        print(f"-> {aluno.nome} (Média: {aluno.media():.1f})")

print("\n--- Aluno com Maior Média ---")
if aluno_maior_media:
    print(f"Nome: {aluno_maior_media.nome} com média {maior_media:.1f}")
````
