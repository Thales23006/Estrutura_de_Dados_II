ATIVIDADE 1

Crie um programa que:

crie uma lista com 10 números inteiros;
solicite os valores ao usuário;
exiba todos os números;
calcule a soma;
calcule a média.
Desafio
Além disso, encontre:

o maior valor;
o menor valor;
a quantidade de números pares.

```
# Isto` está formatado como código`
```



````python
# LISTA
valores = []

for valor in range(10):
    valor = int(input(f"Digite o {valor+1}° numero inteiro: \n"))
    valores.append(valor)
    print(valores)


soma = sum(valores)
print(soma)
media = soma/ len(valores)
print(media)
print(max(valores))
print(min(valores))

pares = 0

for valor in valores:
  if valor % 2 == 0:
    pares+= 1






print(f"Pares {pares}")



````
